# Exercícios de Pandas com Base de Dados do Spotify

```python
import pandas as pd
import kagglehub

# Baixando a última versão no Kaggle
path = kagglehub.dataset_download("maharshipandya/-spotify-tracks-dataset")

# Leitura direta
df = pd.read_csv(path + "/dataset.csv")

# Selecionando as colunas desejadas
df = df[
    [
        "track_name",
        "artists",
        "track_genre",
        "popularity",
        "duration_ms",
        "danceability",
        "energy",
        "key",
        "loudness",
        "mode",
        "speechiness",
        "acousticness",
        "instrumentalness",
        "liveness",
        "valence",
        "tempo",
        "time_signature"
    ]
]

# Renomeando nomes
df = df.rename(columns={
    "artists": "artist",
    "track_genre": "genre"
})

print(df.head())
```


1) A plataforma Spotify deseja identificar quais músicas possuem maior destaque entre os usuários. Utilize recursos do pandas: seleção de colunas, filtragem booleana, `sort_values()` e `head()`. Utilizando a base de dados fornecida selecione apenas as colunas: `track_name`, `artist`, `genre` e `popularity`. Filtre apenas as músicas com popularidade maior que 80 e apresente os resultados ordenados da maior para a menor popularidade.




```python
# Selecionando apenas as colunas solicitadas, chamo minha varviável, seleciono meu data frame que foi chamado de "df" e entre colchetes
# Coloco os nomes das colunas que irei trabalha
selecionar_colunas = df[["track_name", "artist", "genre", "popularity"]]

# Filtrando músicas com popularidade maior que 80, seleciona a variável que separou as colunas que queriamos, e aplicamos a condição simples (>80 "maior que oitenta")
selecionar_colunas = selecionar_colunas[selecionar_colunas["popularity"] > 80]

# Ordenando da maior para a menor popularidade, chamo o sort_values e apos selecionar a coluna que quero, aplico o ASCENDING que vai deixar em ordem decrescente
selecionar_colunas.sort_values("popularity", ascending=False)

# Apresentando os primeiros resultados com head (os 5 primeiros resultados)
print(selecionar_colunas.head())
```


2) O Spotify deseja criar categorias para classificar automaticamente o perfil musical das músicas com base nos atributos: `danceability`, `energy` e `valence`. Crie uma nova coluna chamada `perfil_musical`, que com base na média destes atributos, classifique cada música como: `"Calma"` (<0.4), `"Moderada"` (<0.7) e `"Agitada"` (>0.7). Utilize recursos do pandas: `apply()`, criação de funções e criação de novas colunas.


```python
#Resolução da professora
def classificacao (atributos):
  media = (atributos["danceability"] + atributos['energy'] + atributos['valence'])/3
  if media < 0.4:
    return "Calma"
  elif media < 0.7:
    return "Moderada"
  else:
    return "Agitada"

df["perfil_musical"] = df.apply(classificacao, axis=1)

df[["track_name", "perfil_musical"]].head()
```

3) O Spotify deseja analisar a evolução dos lançamentos musicais ao longo do tempo. Como a base de dados não possui uma coluna de datas reais de lançamento, crie uma coluna fictícia chamada `release_date` utilizando datas sequenciais. Utilize recursos do pandas: `pd.date_range()`, `pd.to_datetime()`, `indexação temporal`, `slicing temporal`, `loc[]` e `sort_index()`. Em seguida:
- converta a coluna release_date para o formato de data;
- defina essa coluna como índice temporal;
- selecione apenas músicas lançadas em 2020;
- filtre músicas lançadas entre 2018 e 2021;
- apresente os resultados em ordem cronológica.


```python
import pandas as pd

# 1. Criar a coluna fictícia release_date com intervalo em minutos ('min')
# Com 'min', 114.000 registros avançam apenas ~79 dias a partir de 2018
df['release_date'] = pd.date_range(start='2018-01-01', periods=len(df), freq='min')

# 2. Converter a coluna release_date para o formato de data (datetime)
df['release_date'] = pd.to_datetime(df['release_date'])

# 3. Definir a coluna release_date como índice temporal
df = df.set_index('release_date')

# 4. Ordenar o índice temporal para garantir a ordem cronológica
df = df.sort_index()

# 5. Selecionar apenas músicas lançadas em 2018 (ou no ano desejado) usando .loc[]
musicas_2018 = df.loc['2018']
print("--- Músicas lançadas em 2018 ---")
print(musicas_2018.head())

# 6. Filtrar músicas lançadas entre 2018 e 2021 em ordem cronológica
musicas_2018_2021 = df.loc['2018':'2021'].sort_index()
print("\n--- Músicas lançadas entre 2018 e 2021 (Ordem Cronológica) ---")
print(musicas_2018_2021.head())
```
